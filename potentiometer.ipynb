A few months ago I set out to remove and replace my Duo's surface-mounted capacitors. Sadly, and despite (what I had believed to be) an abundance of caution that entailed generous use of Kapton tape, I managed to melt half of the headphone volume potentiometer's dial wheel:

![Duo Volume Pot](potentiometer_images/duo_vol_pot_m_sz_h_qual.jpeg)

To remedy this mishap, I embarked on what would turn out to be a months-long search for a suitable replacement. Along the way, I noticed that [no small](https://www.pcenginefans.com/forum/index.php?topic=22830.msg511979#msg511979) [number](https://www.pcenginefans.com/forum/index.php?topic=22007.msg487253#msg487253) of [folks](http://kohada.5ch.net/test/read.cgi/retro/1199327933/698-708) on both English and Japanese language forums have also needed to replace this part. So to help fill in this gap in the knowledge base, I'd like to share what I learned during my own extended process of trial and error.

3. This pot's taper (the rate at which volume increases as the wheel gets rotated) is linear, which means it can adjust volume as naturally as possible given the constraints imposed by the design of the Duo's headphone audio circuit.
4. The dial wheel provides a healthy amount of physical resistance when rotated. This makes it harder for the user to accidentally rotate the wheel too far and hurt their ears by making the headphone volume louder than intended.

### Runner-Up Replacement Candidate
An acceptable alternative is [this 10KΩ potentiometer](https://www.radioshack.com/products/radioshack-10k-ohm-wheel-potentiometer) sold by Radioshack. ### Worst Replacement Candidate
Under no circumstances should you *ever* use the [Alps RK10J12R0A0B](https://tech.alpsalpine.com/prod/e/html/potentiometer/rotarypotentiometers/rkj1/rk10j12r0a0b.html) as a replacement potentiometer for the Duo! 

I can't emphasize this warning enough because I believe that most audio experts, after reviewing this pot's characteristics, would incorrectly jump to the conclusion that it is the ideal replacement pot.

Indeed, Alps' [datasheet](https://www.mouser.com/datasheet/2/15/RKJ1-1370680.pdf) indicates that the RK10J12R0A0B pot has a logarithmic taper, and generally speaking, log tapers do allow for a more natural volume adjustment than linear tapers. However, as I'll explain below, owing to some peculiarities of the Duo's circuit design, I can promise you that logarithmic will turn out to be *the worst* of all possible taper choices for this console.

### Why Logarithmic Tapers are (Usually) Best
The reason we usually prefer log tapers over linear tapers is becuase, as [this article](electronics-tutorials.ws/resistor/potentiometer.html) on Electronics Tutorials explains, the human ear's sensitivity to changes in volume can best be approximated by a logarithmic function. When using a potentiometer with a log taper to raise volume, even though the bulk of the volume increase will occur toward *the end* of the dial wheel's full rotation, the human ear will perceive that all along volume intensity has been increasing at a constant rate. Conversely, if volume intensity were to actually be increased in a steady, linear fashion, the ear would instead perceive that loudness is increasing too drastically in the early part of the potentiometer's rotation.

For us humans, a change in volume that seems to be happening at a constant rate is what feels most natural. The trick is that in order for us to be able to perceive that a volume increase or decrease is linear, the change actually needs to occur at a rate that varies logarithmically.

### Anti-log Potentiometer Tapers
As [this article](https://sound-au.com/pots.htm#taper) by Rod Elliott explains, nowadays almost all potentiometers have either a linear or logarithmic taper. But once upon a time, manufacturers also used to produce pots that were anti-logarithmic. It's crucial to recognize that an anti-log pot would have offered the worst possible user experience for audio, seeing as how it had a taper that's the *inverse* of the log function.

Though anti-log pots no doubt served some useful purposes of their own, they are now are mostly a thing of the past. Yet even so, we as Duo owners need to be familiar with this very audio-unfriendly taper due to the sad fact that if a logarithmic pot were to be installed in the Duo, the Duo would make it behave as if it had an anti-log taper!

![Pot Tapers](potentiometer_images/pot_tapers_m_qual.jpeg)

### The Duo's Headphone Audio Circuit is Atypical
Consoles like the Game Boy or Game Gear allow volume to rise as the potentiometer's wheel is rotated to decrease the resistance between the pot's input and output pins. This convention is also presumed by the taper drawings found in potentiometer [datasheets](https://www.mouser.com/datasheet/2/15/RKJ1-1370680.pdf), as well as in the taper sketch that I included just above. However, it's critical to note that the PC Engine Duo's headphone audio circuit *does not* interface with its volume potentiometer in the standard way, and that the **the Duo's headphone volume increases as its volume potentiometer's resistance increases!**

The consequence of this is that if we were to install a pot with a logarithmic taper, the Duo would travel backwards from the top-right, down and to the left along this pot's "standard" resistance taper curve as volume increases. Volume would initially skyrocket and then slowly plateau, and such a potentiometer would for all intents and purposes appear to the Duo to have an anti-log taper. In this scenario, the user is liable to blow out their ears because the Duo's headphone volume would instantaneously max out after the pot's dial wheel is rotated just a tiny fraction of its full rotation. 

I learned all of this the hard way after I tried installing the Alps RK10J12R0A0B inside my own Duo and found that it makes for horribly unnatural headphone volume adjustment. [This individual](https://www.pcenginefans.com/forum/index.php?topic=22830.msg512366#msg512366) also independently discovered the same thing. 

Once again, whatever you do, do not install in the PC Engine Duo this particular pot or any pot whose datasheet describes it as having a logarithmic taper!

### Why the Duo Does Things Differently
I surmise that the two simple constraints of circuit board design and consumer convention forced the Duo's creators at NEC to design a headphone circuit where increasing the volume pot's resistance makes the volume get louder:
1. It was most convenient to place the headphone audio circuit on the top of the Duo's double-sided circuit board.
2. The Duo needed to adhere to the consumer electronics convention of having headphone volume increase as the volume dial is turned *to the right*.

#### The Duo's Circuit Board is *Crowded*
It would be a gross understatement to say that the PC Engine Duo's two-sided PCB has a lot packed into it. The underside contains several chips that handle graphics and manage the CD-ROM drive:
![PCB Bottom](potentiometer_images/pcb_bottom_m_sz_h_qual.jpeg)

Conversely, the bulk of the audio circuitry resides in the upper-left quadrant on the top of the Duo board:
![PCB Top](potentiometer_images/pcb_top_m_sz_h_qual.jpeg)

My hunch is that as the NEC engineers negotiated their way through the various design conflicts that surely cropped up as they jam-packed all of the Duo's circuitry onto its two-sided PCB, they ultimately concluded that it would be expedient to place the volume potentiometer on the upper side of the board along with all the other audio wiring.

#### Consumer Electronics Convention: To Raise Volume, Turn Right
However, placing the volume potentiometer on the top of the PCB would immediately put the Duo at risk of clashing with convention because:
1. Most dial wheel potentiometers increase resistance as the wheel is rotated *counter-clockwise*.
2. Most headphone circuits are designed to have volume increase as the pot's resistance is lowered (wheel turned *clockwise*).
3. Consumer products with a headphone jack usually have volume increase as the wheel is twisted rightward.

### How the Duo's Headphone Circuit is Wired
Note that the Game Boy and Game Boy Color avoid this connundrum because they both place the volume pot on the underside of their circuit board:
![Game Boy Color PCB underside](potentiometer_images/gameboycolor_board_bottom_m_sz_m_qual.jpeg)
*Image source: [Wikipedia](https://upload.wikimedia.org/wikipedia/commons/8/89/Nintendo-Game-Boy-Color-Motherboard-Bottom.jpg)*

Placing the volume potentiometer on the underside of the PCB means that turning the volume wheel to the right, from the user's perspective, also corresponds to a *clockwise* rotation of the volume pot, which in turn lowers the potentiometer's resistance, which raises the volume.

In other words, because the Game Boy Color's volume pot goes on the bottom side of its circuit board, it can both adhere to popular consumer electronics convention and be wired into the audio circuit in the standard way:
![Game Boy Color Volume Pot Schematic](potentiometer_images/gameboycolor_vr1_schematic.png)
*Image source: [Console5](https://console5.com/wiki/File:Nintendo_GBC_Schematic.png)*

In following this standard configuration, the Game Boy Color wires its volume pot *in-series* with the left and right audio channels. These signals enter the pot through its input pins (one pin per channel) and then depending on how high the pot's resistance is set, they lose a portion of their voltage before they exit through the pot's two output pins before continuing their journey onto the amplifier and then the headphone port. In this way, the GBC pot acts like a nightclub bouncer that only lets a certain amount of audio signal strength (voltage) in through the door.

Because the PC Engine Duo's designers placed its volume pot on the topside of its PCB, turning the volume wheel to the right from the user's perspective corresponds to a *counter-clockwise* rotation of the pot, which raises the the pot's resistance. And so the only way for NEC to get the Duo to adhere to popular consumer convention where a rightward turn of the volume wheel raises headphone volume was for the engineers to design the following audio circuit:
![PC Engine Duo Volume Pot Schematic](potentiometer_images/duo_vr601_schematic.png)
*Image source: [Furrtek](https://twitter.com/furrtek/status/1274369909015838720) and [Console5](https://console5.com/wiki/File:Turbo-Duo-Audio-Schematic.png)*

Here's how the above schematic looks in real-life on a Duo PCB:
![Duo Headphone Audio PCB](potentiometer_images/duo_headphone_pcb_l_sz_h_qual.jpeg)

As this [post explains](https://www.pcenginefans.com/forum/index.php?topic=960.msg113719#msg113719), the Duo's volume potentiometer is not placed in-series with the left and right audio channels. Instead, each of the pot's two input pins is wired *across* one of the circuit's left and right audio channels and the pot's two output pins are connected to the ground plane. When the pot's resistance is set as low as possible, the audio signals lose the maximum possible amount of strength as the majority of their voltage becomes shorted out to ground. When this happens, headphone volume is at a minimum. As the pot's resistance is raised, a correspondingly smaller fraction of the two channels's voltage is diverted off to the ground plane through the potentiometer. Volume hits its maximum at the point where the pot's resistance is as high as possible.

In contrast to a nightclub bouncer that only lets some fraction of the left/right audio voltage proceed through the nightclub entrance and onto the headphone output, we can think of the Duo pot as being like a pipe valve that's connected to the sides of two water mains that carry the left and right sound channels. When the pipe valve is open wider (pot resistance is low), more of the audio voltage leaves the main pipes and gets sent out through those side pipes, is dumped in a big pool (the ground plane), and *doesn't ever* make it to the headphone jack. As the pipe valve gets closed (resistance is increased), less voltage gets diverted and so a stronger (louder) audio signal is allowed to proceed onward to the amplifier and headphones.

The PC Engine Duo's designers understood that their headphone audio circuit design would make a logarithmic pot behave like an anti-log pot. #### Sidenote: "Resistance Taper" vs. "Audio Taper" vs. "Taper"
Throughout this write-up I use the terms "resistance taper," "audio taper," and "taper" interchangeably to refer to the same thing: the rate at which headphone volume increases as the pot's dial wheel gets rotated. 

In most consumer electronics, volume goes up as the pot moves *down* its resistance taper from high to low resistance. Recall, however, that for the Duo headphone volume increases as the volume pot moves *up* its resistance taper, increasing the resistance between its input and output pins. Note that because of this positive correlation, in the remainder of this writeup I will use the rate at which a pot's resistance increases to serve as a proxy for the rate at which perceptible headphone volume rises.

### Identifying Compatible Pin-outs
It's very unlikely you'll come across a 5-pin stereo dial wheel potentiometer with a pin-out that isn't compatible with the Duo's circuit design. The only criteria that matter are:
1. Pins 5 and 4 must receive the stereo audio signals as inputs.
2. Pins 3, 2, and 1 must hold the two stereo output pins as well as the pot circuit's ground pin. Since these are all ultimately are connected to the Duo's ground plane, it doesn't matter which of these three pins is responsible the L/R channel outputs and pot ground.

For example, the Game Boy's volume potentiometer would be compatible with the Duo's headphone audio circuit even though it [has a slightly different pin-out](https://sudomod.com/forum/viewtopic.php?f=22&t=1280&sid=6f6298f419d0bfc9fb9435aa531035c3&start=10):
* Pin 5: 1st Audio Channel Input
* Pin 4: 2nd Audio Channel Input
* Pin 3: 2nd Audio Channel Output
* Pin 2: 1st Audio Channel Output
* Pin 1: Ground

Additionally, I've [seen it mentioned](https://www.pcenginefans.com/forum/index.php?topic=22830.msg512121#msg512121) that the Game Gear's volume pot is incompatible with the Duo owing "a different pin-out." This is absolutely incorrect. While the Game Gear pot's pin-out is indeed different from the stock Duo's, it is actually identical to that of the Game Boy pot and it meets the two compatibility criteria:
![Game Gear Pin-out](potentiometer_images/game_gear_pot_pinout_l_sz_m_qual.jpeg)

Basically, as long as the pot has 5 pins and the stereo channels enter the pot through pins 5 and 4, the pin-out will be compatible.

There do exist 5-pin pots with incompatible pin-outs, but it's far more likely that physical constraints like wheel diameter and wheel height will be the most meaningful way to filter candidates. Incidentally, neither the Game Boy nor Game Gear volume pots do well in this regard, as each has a wheel height that's far too tall to fit through the narrow opening in the Duo's chassis.

### Pursuing Perfection: Does Maximum Resistance Matter?
At this point, you may have noticed that while the Duo's stock potentiometer has a max resistance of 20KΩ, the Alps pot I recommended offers a top resistance of only 10KΩ. Now if you're a perfectionist like me, you might be asking yourself whether it would also be useful to search for a replacement pot that offers a higher maximum resistance than the Alps pot's 10KΩ.

The good news here is that for most use-cases, the replacement pot's maximum resistance *doesn't matter*. However, there are a small number of scenarios where maximum resistance *might* matter. For example, if the user wanted to connect external stereo speakers (not headphones) to their Duo's stereo headphone jack and then planned to turn the Duo's volume wheel up quite high, a pot with a higher max resistance would ensure that the sound quality is still crisp, clear, and defined even at this very loud volume.

Nonetheless, since the Duo's A/V port also outputs stereo audio, I believe that very few folks would wish to do this, and so I stand by my assertion that most of us don't need to worry about a replacement potentiometer's total resistance.

### Maximum Resistance (Usually) Doesn't Matter
Based on what I've read online, however, it appears that more than a few individuals are under the incorrect impression that maximum resistance is a crucial consideration. My hunch is that this is because most folks in the PC Engine community have an incorrect mental model of how the Duo's volume potentiometer actually attenuates (aka. adjusts or lowers) the strength of the console's headphone audio signal. They mistakenly assume that the Duo behaves similar to the Game Boy and Game Boy Color, where rotating the volume wheel to increase the potentiometer's resistance leads to quieter audio. However, recall that as I explained above, PC Engine Duo headphone volume is at its lowest when its volume potentiometer's resistance is *also at a minimum*.

An example of this misunderstanding can be found inside [this post](https://misukou.blog.fc2.com/blog-entry-3152.html) by a Japanese modder that explains how to replace the Duo's volume pot with a donor from the Game Boy Color. All things being equal, it's a useful and accessible write-up that demonstrates that a GBC volume pot is compatible with the Duo's PCB and plastic chassis. However, at the end of this post the modder acknowledges that rotating the Game Boy Color donor pot beyond just a few degrees will result in an extremely uncomfortably loud volume level. The writer's mistake in all of this is that they presume that this is due to the fact that the GBC donor pot's maximum resistance of 10KΩ is only half that of the Duo's stock volume potentiometer.

First off, it is true that the GBC volume pot's max resistance is only half that of the original Duo pot. However, given the fact that with the Duo, *minimum* volume happens at *minimum* resistance, why on earth should we expect a volume pot's *maximum* resistance level to have any influence on that pot's ability to keep the Duo's minimum headphone volume as quiet as possible?

Furthermore, as for how things proceed once the user begins to raise volume from its minimum, like I explained above, a volume potentiometer's ability to enable gentle and natural sound adjustment depends solely on the potentiometer's taper, which is the rate at which audio signal strength varies as the pot is rotated. 

### The Duo Volume Pot is a Voltage Divider
So I've made the intuitive argument that for the PC Engine Duo, there's no logical reason to assume that maximum resistance has any relation to a potentiometer's ability to keep volume from getting too loud too fast. However, you may also be surprised to learn that even for consoles like the Game Boy that install the volume pot in the standard way, maximum resistance *also doesn't matter*.

Many folks, including a prior version of myself, tend to imagine that video game volume potentiometers function as rheostats (variable resistors). We incorrectly presume that rotating a volume pot's dial wheel changes the value of "the resistor" that is buried somewhere deep inside the potentiometer. While there are potentiometers that work like this, this is not how these pots work. Instead, the volume pots found in most home consoles use a circuit that takes a single resistor and divides it so that it behaves like two separate resistors. Using the dial wheel to adjust the relative sizes of these two resistors will in turn raise or lower the voltage of the current found at the pot's output pins. The size of this voltage is what determines the loudness of the audio produced by the headphones.

It's easier to understand how all of this works if we take a look at how a potentiometer makes use of its physical components to accomplish this:

A stereo pot has a carbon resistive strip for each of the two audio channels (as evidenced by the following specimen I took from a Game Boy volume pot):
![Game Boy Vol Pot Carbon Resistive Strip](potentiometer_images/gameboy_pot_resistive_strip_l_sz_m_qual.jpeg)

The underside of the dial wheel touches these strips in such a way that effectively splits each strip into two separate resistors:
![How a Potentiometer Works](potentiometer_images/how_dial_wheel_pot_works_l_sz_h_qual.jpeg)

As the dial wheel gets rotated, the values of the respective resistances that belong to these two "resistors" begin to change. And it is the relative values of these two resistances, and nothing else, that is the sole determinant of of the fraction of the input voltage that gets diverted to become the output voltage of each audio channel. This kind of circuit is known as a voltage divider. 

Here is a concise schematic that represents the voltage divider circuit that belongs to one of the audio channels in a stereo potentiometer:
![Simple 1-Channel Volume Pot Schematic](potentiometer_images/simple_1_channel_pot_schematic_m_qual.jpeg)
*Schematic originally from the Alps RK10J series' [datasheet](https://www.mouser.com/datasheet/2/15/RKJ1-1370680.pdf)*

And here is a more detailed representation of the voltage dividing circuit formed when a volume pot's dial wheel is rotated to an angle that splits the entire resistive strip, $R_{total}$, into two smaller resistors $R_{1}$ and $R_{2}$:
![Voltage Divider Circuit](potentiometer_images/voltage_divider_circuit_m_qual.jpeg)
*Schematic originally from [this](https://www.allaboutcircuits.com/tools/voltage-divider-calculator/) All About Circuits article*

The output voltage of the current found at terminal 2 is determined by the equation:
$$
V_{out} = V_{in} * \frac{R_{2}}{R_{1} + R_{2}}
$$

As a potentiometer's dial wheel is rotated counter-clockwise, the value of R1 will increase and the value of R2 will decrease. The equation for $V_{out}$ indicates that as this happens, the output voltage at terminal 2 will decrease. 

For a given pot, no matter how $R_{1}$ and $R_{2}$ may vary relative to each other, the pot's maximum resistance, $R_{total}$, remains constant as it's just the sum of $R_{1}$ and $R_{2}$. Or to put it another way, regardless of the size of $R_{total}$ it's always true that $R_{2}=R_{total}-R_{1}$, and so the *relative* sizes of $R_{1}$ and $R_{2}$ will never be contingent whatever absolute value $R_{total}$ happens to be.

In mathematical terms, the range of values of the expression $\frac{R_{2}}{R_{1} + R_{2}}$ will always be the same regardless of $R_{total}$. The implication of this is that, theoretically speaking at least, a potentiometer's total resistance has no bearing on how the voltage of the output signal varies as the pot gets adjusted.

#### Clarifying What "Increasing Resistance" Means
If you'll recall, earlier in this write-up I explained that the Duo's headphone volume increased as its volume pot's resistance also "increased." This may strike some readers as being contradictory with my more recent assertion that the Duo volume pot is not a rheostat (variable resistor) but a voltage divider. 

Whenever I use as a shorthand the language of increasing a pot's "resistance," I am referring to the measured resistance between the input and output pins of one of the pot's stereo channels. In the context of the potentiometer's voltage divider circuit, increasing this "measured resistance" is the same thing as "increasing the value of $R_{1}$."

### Graphing $V_{out}$
If reviewing the above equation hasn't quite made things click, here are a few graphs to drive home the point that for a given incoming voltage, $V_{in}$, the outgoing voltage, $V_{out}$, of a potentiometer's voltage divider circuit depends exclusively on the size of $R_{2}$ *relative to* that of $R_{1}$, *and not* on the pot's maximum resistance $R_{total}$.

Let's pretend we have a circuit that has a $V_{in}$ of 5V, and that we've connected a potentiometer with a maximum resistance $R_{total}$ of 20KΩ. Here's how $V_{out}$ would vary as the pot's dial wheel gets rotated in a counter-clockwise direction that increases $R_{1}$ and simultaneously decreases $R_{2}$:

````python
def get_Vout(V_in, R_total):
    R_one = list(range(R_total+1))
    R_two = [R_total - i for i in R_one]
    V_out = [V_in*(R_two[i]/(R_one[i]+R_two[i])) for i in range(R_total+1)]
    return V_out, R_one
````

````python
V_in = 5
R_total = 20

fig, ax = plt.subplots(nrows=1, ncols=1, figsize=(10, 8)) 
def plot_Vout(V_in, R_total, ax):
    title = '$V_{out}$ when a $R_{total}$=' + f'{R_total} KΩ Pot is Turned Counter-Clockwise'
    V_out, R_one = get_Vout(V_in, R_total)
    ax.set_xlabel('$R_{1}$ (in KΩ)')
    ax.set_ylabel('Output Voltage\n$V_{out}$ (in V)', rotation=0)
    ax.yaxis.set_label_coords(-0.16,0.5)
    ax.set_title(title, pad=15)
    ax.plot(R_one, V_out)
    secax = ax.secondary_xaxis('top', functions=(lambda x: R_total-x, lambda y: R_total-y))
    secax.set_xlabel('$R_{2}$ (in KΩ)')
plot_Vout(V_in, R_total, ax)
plt.tight_layout(pad=5)
plt.show();
````

(Note that this graph is not displaying our hypothetical potentiometer's "taper"! This is because the graph doesn't capture the *rate* of increase or decrease of resistance as the pot gets rotated to a given angle. Instead, the plot simply details the simple linear relationship between $V_{out}$ and $V_{in}$ and the resistances $R_{1}$ and $R_{2}$.)

Finally, let's prove to ourselves that this relationship between the relative sizes of $R_{2}$ and $R_{1}$ in the function that governs output voltage, $V_{out} = V_{in} * \frac{R_{2}}{R_{1} + R_{2}}$, remains the same regardless of our hypothetical pot's total resistance $R_{total}$:

````python
resistances = [[5, 10], [20, 50]]
nrows, ncols = 2, 2
fig, axs = plt.subplots(nrows=nrows, ncols=ncols, figsize=(14, 12))
for r in range(nrows):
    for c in range(ncols):
        plot_Vout(V_in, resistances[r][c], axs[r][c])
plt.tight_layout(pad=5)
plt.show();
````

### Sometimes, a Volume Pot's Maximum Resistance Does Matter
The graphs above should have convinced us that when rotating a pot's dial wheel to adjust the fraction of the incoming voltage that gets output from the pot, all we really care about are the *relative sizes* of the resistances of $R_{1}$ and $R_{2}$. Theoretically speaking, the potentiometer's maximum resistance $R_{total}$ does not determine what $V_{out}$ will be.

However, owing to the manner in which dial wheel potentiometers that have carbon resistive strips are constructed, it is not possible to rotate the dial far enough counter-clockwise to make it so that $R_{2} = 0$. At the dial's maximum angle of counter-clockwise rotation, $R_{2}$ will be very small, but it will never be zero. If we think in terms of the measured resistance between a potentiometer's input and output pins, even when the potentiometer is rotated counter-clockwise to make this resistance as large as possible, there will still be some current with very low voltage that escapes from the pot's output pin!

A similar dynamic is also true when a pot's wheel has been fully rotated clockwise to the position where the measured resistance between its input and output pins is at a minimum. Even at this point, it will never be possible to make $R_{1}$ completely zero. Even at "zero resistance" there will be some small amount of resistance faced by the current entering the pot. And because of this, it's not possible to pass on 100\% of the incoming current's voltage to the current exiting from the pot's output pin.

### Residual Resistance
These areas on the far clockwise and far counter-clockwise ends of a pot's carbon strip that are beyond the rotation of a potentiometer's dial wheel are known as **residual resistances**:
![Residual Resistance in the Duo Pot](potentiometer_images/residual_resistance_l_sz_h_qual.jpeg)

Most of the time, the presence of these residual resistances will not affect user experience. Nonetheless, it's important to note that when a pot's maximum resistance (aka. the total resistance of one of its carbon strips) is larger, the pot's two resisidual resistances will *also be larger*. 

Because of this, there are two situations in which a pot's maximum resistance $R_{total}$ *will* actually influence how we experience the audio signal that's attenuated by a volume pot.

### When Maximum Resistance Matters: Scenario 1 - Maximum Attenuation
Minimum headphone volume will be lower if we use a pot with a *lower* max resistance. 

When the PC Engine Duo volume pot's dial wheel is rotated fully clockwise (to minimum resistance between its input and output pins) to make volume as low as possible, there will always be some small smidgen of the carbon resistive strip that remains inside $R_{1}$, and thus it won't be possible to ground out *all* of of the audio signal. When a potentiometer has a larger total resistance, this "smidgen" of the carbon strip that always stays in $R_{1}$ (the residual resistance) will also have a larger resistance in absolute terms. 

The key thing to understand is that if this residual resistance is larger, the pot will ground out *less* of the audio signal's voltage at minimum volume. This is because the larger residual resistance is preventing a greater amount of voltage from heading out to the ground plane. This [wonderful article](https://tech.alpsalpine.com/e/products/faq/potentiometer/features.html) on the Alps website explains that a potentiometer's ability to make volume low as low as possible is known as its **maximum attenuation**. 

A pot with a larger residual resistance will always provide a *lower* maximum attenuation than a pot with smaller residual resistance. This is good news for those of us replacing the Duo's original 20KΩ pot with a 10KΩ version. By choosing a replacement pot with lower total resistance, we are actually improving on the original experience because the new pot will make the minimum headphone volume quieter!

(Note that when a volume potentiometer is wired to an audio circuit in the standard way, as is the case with the Game Boy and Game Boy Color, maximum attentuation happens when the pot's resistance is *as high as possible*. This is what's presumed by the Alps article I just referenced. However, remember that because the PC Engine Duo does things backwards, maximum attenuation happens when the volume pot's resistance is turned down as low as it'll go.)


### When Maximum Resistance Matters: Scenario 2 - Insertion Loss
The sound quality at maximum headphone volume will be better defined (less muddy-sounding) if we use a pot with a *greater* max resistance.

Analagous to the residual resistance zone that determines a pot's maximum attenuation, every dial wheel potentiometer also has a second area of residual resistance lying at the opposite end of its carbon resistive strip. In the same way that the size the first residual resistance determines *how low* the potentiometer can make the volume (maximum attenuation), this second zone of residual resistance defines *how high* the pot will allow volume to go. This property is called **insertion loss**.

For any potentiometer installed in the PC Engine Duo, the zone of residual resistance that determines the size of the pot's insertion loss is found on the end of the carbon strip that lies just beyond the dial wheel's angle of full *counter-clockwise* rotation. This would be the point at which the resistance between the pot's input and out pins is highest, and at which headphone volume is loudest (because the largest possible portion of the audio signal's voltage is prevented from being shorted to the ground plane).

Recall that as far as user experience is concerned, we wanted maximum attenuation to be as large as possible so that the minimum volume could be as quiet as possible, and that this could be achieved by installing a potentiometer with a lower total resistance. Again, the reason for this is that a pot with lower total resistance can output a weaker "minimum strength" signal (having a lower voltage) than could a pot with a higher total resistance.

On the other end of the spectrum, the Duo's "maximum strength" audio will sound better if it has a relatively lower voltage. If the voltage of the audio signal sent out from the headphone jack at "full volume" is too high, it is unlikely that the connected speakers will be able to produce a clear sound that conveys the full definition of the audio track. When the audio starts to sound muffled or muddy as volume is raised, this is likely the reason why.

In contrast to the prescription for ideal maximum attenuation, the best way to prevent prevent a muddy/muffled sound at max volume is to install a volume potentiometer that has a *higher* total resistance. Such a potentiometer will have a larger insertion loss, which will prevent the voltage of the audio signal at loudest volume from getting too high. This will keep sound quality clear and defined even as the pot gets rotated to full volume.

### Insertion Loss Matters More to the User Experience
The Duo's original designers decided that ensuring clear, defined audio even at *very high* volume was more important to the user experience than having minimum volume be as quiet as possible. I believe this is the reason that the stock Duo volume pot is 20KΩ. 

Unfortunately for us, 20KΩ stereo pots with compatible wheel dimensions are not widely available today and the 10KΩ Alps pot I recommended as a replacement does indeed produce muddy audio if it is turned up all the way. The saving grace, as I'll explain further below, is that throughout the entire volume range that's comfortable to the human ear when heard via headphones, my favored Alps pot keeps audio crisp, clear, and defined. 

Seeing as how the Duo's A/V port also outputs stereo audio, I believe that almost anyone using the console's headphone jack will be doing so via headphones rather than external speakers. As such, I expect that almost nobody would ever have reason to care about audio quality at the Duo pot's "full volume," and so I'm confident that virtually all Duo owners would be more than pleased with the 10KΩ Alps that I recommend as the ideal replacement pot for the system.

### Pursuing Perfection: The Dream Scenario Potentiometer for the PC Engine Duo
Thus far, I've explained why the 10KΩ [Alps RK10J12E002L](https://tech.alpsalpine.com/prod/e/html/potentiometer/rotarypotentiometers/rkj1/rk10j12e002l.html) potentiometer is the ideal replacement candidate for the Duo thanks to its size and linear taper. I also revealed that logarithmic taper belonging to the Alps RK10J12R0A0B makes it the worst candidate. Unlike normal consoles like the Game Boy or Game Boy Color, peculiarities in the Duo's design mean that a pot with a log taper will behave as if it had an anti-log taper! The Duo's designers were aware of this and that's why the system's original pot had a linear taper. An anti-log pot would have been best (because the Duo would have made it behave as if it had a logarithmic taper), but even as long ago as the early '90s such pots were apparently not available from commodity parts suppliers!

I then broke down the tradeoffs that arise when deciding how high the maximum resistance of the replacement pot should be. A pot with a relatively lower maximum resistance like 10KΩ will enable the minimum headphone volume *to be lower* than what it would be if a pot with higher maximum resistance were chosen. On the other hand, it's also easy to see why the Duo's engineers originally chose a volume pot with 20KΩ maximum resistance. Doing so meant that sound quality at maximum volume was much better defined and less muddy than what would have been the case had a lower resistance pot been used.

I agree with the Duo's engineers that keeping loud volume as clear as possible is a far more important consideration than keeping minimum headphone volume as muted as possible. Therefore, here are the specs that I believe the perfect PC Engine Duo volume pot would possess:

1. Wheel diameter 14mm
2. Wheel height 9mm
3. 2-gang (channel) stereo circuit, with 5 pins and a compatible pin-out
4. Anti-log taper (which the Duo would treat like a log taper)
5. 20KΩ total resistance
6. Wheel has just the right amount of physical "push-back" when rotated (so that it doesn't spin too easily)

### Unfortunately, the Perfect Potentiometer for the Duo Doesn't Exist
Right off the bat, we must acknowledge that an anti-log taper is a non-starter. They simply don't exist. Furthermore, I've been unable to locate any 20KΩ or 50KΩ pots that have compatible pin-outs and wheel dimensions. I could not find a single 20KΩ pot that had 5 pins, and all 50KΩ pots I came by had wheel heights at least 2mm thick - far too thick for the Duo. (I did find one 50KΩ pot from Alps with ideal wheel dimensions, but it used pin 3 as the ground pin, which simply won't work with the Duo.) 

### Comparing Tapers and "Comfortable Audio Ranges"
I like the Alps RK10J12E002L because it meets more of the above six criteria than any other pot that's commercially available today. However, if we could one day find a pot with the right wheel dimensions *and* a gentler linear taper, this would definitely provide a better user experience. The implication of having a gentler taper (one with a flatter slope) is that a larger percentage of the pot's full rotation angle range would output audio that sounds comfortable. Such a potentiometer would feel like it had "more play" - it'd be less likely for the user to accidentally make the volume too low or too high. 

To sum things up: the gentler the potentiometer's taper, the easier it is to get a headphone volume that sounds "just right." The range of dial wheel rotation angles throughout which audio is neither too high nor too low is what I call the "comfortable audio range." And I believe that this is the next most important criteria by which to judge a possible replacement pot that meets all other non-negotiables like wheel dimensions and pin-out.

This is because the Duo's original pot could rotate a full 300°, while all other pots would rotate a maximum of 270°. As far as the y-axis is concerned, units are standardized to % of max resistance.
````python
# Resistance tapers for the following four pots. All in KΩ.
resistances = [[
    # Stock Duo Pot
    [0.476, 0.253, 0.302, 0.484, 0.79, 1.135, 1.483, 1.642, 2.62, 4.43, 4.83, 5.98, 6.94, 8.06, 9.11, 10.19, 
     11.24, 12.39, 13.29, 14.45, 15.26, 16.36, 17.69, 17.97, 18.38, 18.47, 18.66, 19.05, 18.93, 19.05, 18.98],
    # Alps RK10J12E002L
    [0.0104, 0.0115, 0.0637, 0.1568, 0.335, 0.508, 0.824, 1.28, 1.86, 2.43, 3.01, 3.62, 4.28, 5.05, 5.77, 6.51, 
     7.08, 7.76, 8.45, 9.05, 9.65, 10.19, 10.46, 10.62, 10.72, 10.81, 10.86, 10.85]], [
    # Radio Shack 
    [0.0044, 0.088, 0.248, 0.244, 0.38, 0.51, 0.772, 1.58, 2.37, 3.03, 3.57, 4.29, 4.86, 5.48, 5.92, 6.53, 7.14, 
     7.63, 8.16, 8.92, 9.52, 10.06, 10.3, 10.57, 10.64, 10.69, 10.74, 10.64],
    # Alps RK10J12R0A0B (Measured from the Alps RK10J12E0A0A, which is identical to the RK10J12R0A0B 
    #                    in taper (15A log), max resistance (10KΩ), and pin-out.)
    [0.0068, 0.0143, 0.29, 1.072, 1.911, 2.68, 3.52, 4.35, 5.16, 5.97, 6.9, 7.75, 8.29, 8.48, 8.68, 8.86, 9.05,
     9.23, 9.45, 9.66, 9.87, 10.04, 10.21, 10.32, 10.37, 10.43, 10.47, 10.48]]]

angles = [[list(range(0, len(j)*10, 10)) for j in i] for i in resistances] # in degrees

comfortable_audio_ranges = [[ # in degrees
    # Stock Duo Pot
    (10, 60),
    # Alps RK10J12E002L
    (20, 60)], [
    # Radio Shack 
    (30, 60),
    # Alps RK10J12R0A0B
    (15, 30)]]

titles = [['Stock Duo Pot', 'Alps RK10J12E002L'],
          ['Radio Shack Pot', 'Alps RK10J12R0A0B']]
````

````python
def plot_taper(resistances, angles, comfortable_audio_range, ax, title):
    percent_max_volume = [r*100/max(resistances) for r in resistances]
    percent_full_rotation = [a*100/angles[-1] for a in angles]
    rng = tuple(i*100/angles[-1] for i in comfortable_audio_range)
    
    ax.set_xlabel('% of Full Rotation Counter-Clockwise')
    ax.xaxis.set_label_coords(0.5,-0.1)
    ax.set_ylabel('% of Max\nResistance\n\n(↑Resistance\n⇒↑Volume)', rotation=0)
    ax.yaxis.set_label_coords(-0.17,0.5)
    ax.set_title(title, pad=15)
    ax.axvspan(rng[0], rng[1], color='green', alpha=0.5)
    ax.text(rng[0]+(rng[1]-rng[0])/2, 50, 'Comfortable Audio Range', {'ha': 'center', 'va': 'center'}, 
            color='#006400', font = {'size': 12}, rotation=270)
    ax.plot(percent_full_rotation, percent_max_volume)
````

````python
nrows, ncols = 2, 2
fig, axs = plt.subplots(nrows=nrows, ncols=ncols, figsize=(14, 12))
for r in range(nrows):
    for c in range(ncols):
        plot_taper(resistances[r][c], angles[r][c], comfortable_audio_ranges[r][c], axs[r][c], titles[r][c])
plt.tight_layout(pad=5)
plt.show();
````
